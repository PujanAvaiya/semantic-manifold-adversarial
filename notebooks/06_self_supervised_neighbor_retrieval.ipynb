# 06 — Week 4: Self-Supervised Semantic Neighbor Retrieval

**Goal:** remove ground-truth labels from neighborhood construction and test whether semantic/representation-based retrieval is reliable enough for the geometry.


## Comparison

Compare labeled same-class neighbors with label-free semantic neighbors on semantic purity, neighbor distance structure, local subspace stability, tangent-basis similarity, attack success, distortion, and failure cases.


## Exit criterion

Do not claim a label-independent pipeline unless the label-free neighborhoods are sufficiently stable and useful for the downstream geometry.
