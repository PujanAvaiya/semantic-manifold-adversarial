# 04 — Multi-source attack diagnostics

Run the fixed latent attack on correctly classified MNIST digit-1 examples without modifying its neighbor selection, tangent rank, or update rule.


## Measurements

Per example: source/target confidence, success, optimization steps, latent perturbation norm, decoded distortion, neighborhood distance and label purity, local singular values, explained variance, selected rank `r`, tangent/normal components, decision margin, and boundary crossing.


## Outputs

Per-image CSV, success-rate summary, optimization traces, and representative successful and failed samples. Not yet executed here.
