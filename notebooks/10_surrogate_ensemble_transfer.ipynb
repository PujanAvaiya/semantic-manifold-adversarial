# 10 — Week 7A: Surrogate Ensemble Transfer

**Goal:** determine whether optimizing/evaluating with multiple surrogate classifiers improves generalization beyond a single surrogate.


## Controls

Keep source examples, target classes, perturbation budgets, stopping rules, and representation fixed while comparing single-surrogate and ensemble settings.


## Output

Report per-model and aggregate source success, held-out transfer success, distortion, and any change in direction diversity or stability.
