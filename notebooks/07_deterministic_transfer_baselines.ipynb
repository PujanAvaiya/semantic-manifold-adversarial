# 07 — Deterministic transfer baselines

Measure transfer of the optimized tangent-coordinate perturbation `a*` across independently trained classifiers. Not yet implemented.


## Comparisons

Evaluate new random seeds of the same classifier architecture and at least one different architecture. Keep the VAE, example selection, perturbation budget, and attack procedure fixed.


## Outputs

For each image and model: clean eligibility, source success, transfer success, latent/image norms, reconstruction or semantic checks, and whether the same image or latent direction was transferred.
