# 07 — Week 5A: Deterministic Transfer Baselines

**Goal:** establish what the deterministic optimized tangent-coordinate solution `a*` already transfers before introducing any learned distribution.


## Required tests

Evaluate transfer to independently initialized versions of the same architecture and at least one different victim architecture. Separate source-model success from transfer success. Keep representation/generator changes controlled so the source of transfer is interpretable.


## Report

For each sample/model pair record eligibility, source success, transfer success, perturbation norms, semantic/reconstruction metrics, and whether the same decoded image or the same latent direction is being transferred.
