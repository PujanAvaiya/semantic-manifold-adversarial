# 08 — Local sampling around `a*`

Test whether perturbations sampled around a deterministic tangent-coordinate optimum improve cross-model transfer. Not yet implemented.


## Sampling baseline

Sample `a = a* + noise` across fixed noise scales. Measure source success, transfer success, distortion, sampled-direction diversity, and sensitivity to scale.


## Follow-up condition

Fit a conditional direction distribution only if local sampling reveals a transferable region beyond deterministic `a*`.
