# 05 — Week 3: Adaptive Geometry Ablations

**Goal:** fix only the failure modes demonstrated by Week 2 diagnostics.


## Candidate interventions

Test independently before combining:
- adaptive neighbor selection;
- adaptive tangent dimension `r`;
- target alignment;
- relinearization / tangent-basis re-estimation along the path.

Each intervention needs an explicit diagnostic that motivated it.


## Required decision table

For each observed failure, record the evidence, proposed fix, isolated ablation result, and whether the component should be kept. Components without measurable benefit should not enter the final method.
